```python
from deeplake import Client

load_dotenv()

api_key = os.getenv("DEEPLAKE_API_KEY")
if not api_key:
    raise ValueError(
        "DEEPLAKE_API_KEY is missing. Add it to your .env file or environment."
    )

client = Client(
    token=api_key,
    workspace_id=os.getenv("DEEPLAKE_WORKSPACE"),
)
```
