```python
#==========================================
# Load the data from the website
#==========================================
import bs4
from langchain_community.document_loaders import WebBaseLoader

loader = WebBaseLoader(
    web_path="https://en.wikipedia.org/wiki/Transformer_(machine_learning_model)",
    bs_kwargs=dict(
        parse_only=bs4.SoupStrainer(
            class_=("mw-parser-output",)   # real Wikipedia content class
        )
    ),
)
data = loader.load()
print(data[0].page_content[:500])
```

```python
import bs4
from langchain_community.document_loaders import WebBaseLoader

loader = WebBaseLoader(
    web_path="https://en.wikipedia.org/wiki/Transformer_(machine_learning_model)",
    header_template={
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                      "AppleWebKit/537.36 (KHTML, like Gecko) "
                      "Chrome/120.0 Safari/537.36",
    },
    bs_kwargs=dict(
        parse_only=bs4.SoupStrainer(class_=("mw-parser-output",))
    ),
)

data = loader.load()
print(len(data))                      # should be 1
print(len(data[0].page_content))      # should be thousands of chars
print(data[0].page_content[:300])
```
