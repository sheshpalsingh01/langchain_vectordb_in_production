## Practical examples

### 1. RecursiveCharacterTextSplitter

**Example:** Use it for a long article, research paper, or web page containing paragraphs and headings.

**Do not use it:** When exact token limits are required, because character length does not equal token count.

### 2. CharacterTextSplitter

**Example:** Use it to split a log file where each record is separated by a blank line.

**Do not use it:** When the separator is inconsistent or some sections may be much larger than the desired chunk size.

### 3. TokenTextSplitter

**Example:** Use it when preparing text for an LLM with a strict context-window limit.

**Do not use it:** When preserving complete paragraphs or sentences is more important than exact token control.

### 4. MarkdownHeaderTextSplitter

**Example:** Use it to split a README file into sections such as Installation, Usage, and Configuration.

**Do not use it:** For plain text without Markdown headings.

### 5. RecursiveJsonSplitter

**Example:** Use it to divide a large API response or nested product catalog into smaller structured pieces.

**Do not use it:** For ordinary paragraphs, articles, or other unstructured text.

### Quick decision examples

* A **PDF article** → RecursiveCharacterTextSplitter
- A **newline-separated log** → CharacterTextSplitter
- A prompt with a **strict token limit** → TokenTextSplitter
- A **Markdown README** → MarkdownHeaderTextSplitter
- A large **nested API response** → RecursiveJsonSplitter


```python
from langchain_community.document_loaders import TextLoader, PyPDFDirectoryLoader, UnstructuredMarkdownLoader, UnstructuredWordDocumentLoader,ArxivLoader,WikipediaLoader,YoutubeAudioLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter, CharacterTextSplitter, TokenTextSplitter, MarkdownHeaderTextSplitter, RecursiveJsonSplitter
```

```python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

loader = TextLoader(file_path="data/file.txt")
docs = loader.load()                       # list[Document]

# 1. Recommended general-purpose splitter
recursive_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50,
)

chunks = recursive_splitter.split_documents(docs)   
print(len(chunks))
print(chunks[0].page_content)
```

```python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

loader = TextLoader(file_path="data/file.txt")
docs = loader.load()                       # list[Document]

recursive_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50,
    separators=["\n\n", "\n", " ", ""]
)

chunks = recursive_splitter.split_text(docs[0].page_content) 
chunks
```

```python
loader = PyPDFDirectoryLoader(path='data',glob="*.pdf")
docs = loader.load()
```

```python
Character_Splitter = CharacterTextSplitter(chunk_size = 500, chunk_overlap = 50)
chunk = Character_Splitter.split_documents(docs)
```

```python
# Token Splitting
loader = UnstructuredWordDocumentLoader(file_path='data/docker_setup.docx')
docs = loader.load()

token_splitter =  TokenTextSplitter(chunk_size=100, chunk_overlap = 10)
chunk = token_splitter.split_documents(documents=docs)
```

```python
# Markdown Splitter
loader = UnstructuredMarkdownLoader(file_path='data/Floci_Local_AWS_Cloud_Lab_Guide.md')
docs = loader.load()
markdown_splitter = MarkdownHeaderTextSplitter(
    headers_to_split_on=[
        ("#", "Header 1"),
        ("##", "Header 2"),
        ("###", "Header 3"),
    ]
)
chunk = markdown_splitter.split_text(docs[0].page_content)
chunk
```

```python
# 5. JSON splitter
json_data = {
    "name": "Product",
    "description": "A product description",
    "features": ["Fast", "Reliable", "Affordable"],
}

json_splitter = RecursiveJsonSplitter(max_chunk_size=200)
json_chunks = json_splitter.split_json(json_data)
print(json_chunks)
```
